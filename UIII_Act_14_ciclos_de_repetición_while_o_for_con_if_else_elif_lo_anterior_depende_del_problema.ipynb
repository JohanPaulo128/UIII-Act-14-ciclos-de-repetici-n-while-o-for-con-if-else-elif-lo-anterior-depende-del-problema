### Realizar un algoritmo que pida números (se pedirá por teclado la cantidad de números a introducir). El programa debe informar de cuantos números introducidos son mayores que 0, menores que 0 e iguales a 0.
### Especialidad: Inteligencia Artificial
### Docente: Eliseo Nava


```python
# Datos del alumno
print("Nombre: Johan Paulo Aros Reveles")
print("Número de lista: 04")
print("------------------------------------------")

# Pedir la cantidad de números a procesar
cantidad = int(input("¿Cuántos números vas a introducir?: "))

# Inicialización de contadores
mayores_a_cero = 0
menores_a_cero = 0
iguales_a_cero = 0

# Ciclo para solicitar los números uno por uno
for i in range(cantidad):
    # El mensaje muestra qué número de la serie se está ingresando
    numero = float(input("Introduce el número " + str(i + 1) + ": "))

    # Lógica de comparación
    if numero > 0:
        mayores_a_cero = mayores_a_cero + 1
    elif numero < 0:
        menores_a_cero = menores_a_cero + 1
    else:
        iguales_a_cero = iguales_a_cero + 1

# Mostrar los resultados finales
print("------------------------------------------")
print("Resultados del análisis:")
print("Números mayores que 0:", mayores_a_cero)
print("Números menores que 0:", menores_a_cero)
print("Números iguales a 0:", iguales_a_cero)
```
